## Task 1A - Clean Transactions

The function `clean_transactions(amounts, min_threshold=10.0)` removes:

- non-positive amounts, and
- positive amounts below the minimum threshold.

The assignment specifically requires `filter()` and a `lambda` expression.

```python
def clean_transactions(amounts, min_threshold=10.0):
    cleaned = filter(
        lambda amount: amount > 0 and amount >= min_threshold,
        amounts
    )
    return list(cleaned)
```

## Test `clean_transactions()`

The sample list intentionally contains:

- valid transactions,
- a negative value,
- zero,
- and positive values below the default threshold.

```python
transaction_amounts = [
    250.00,
    -20.00,
    5.00,
    1000.00,
    9.99,
    75.50,
    0.00,
    5000.00,
    15.00
]

print("Original transactions:")
print(transaction_amounts)

cleaned_transactions = clean_transactions(transaction_amounts)

print("\nCleaned transactions:")
print(cleaned_transactions)
```

## Task 1B - Apply Service Fee

The function `apply_service_fee(amounts, fee_rate=0.02)` deducts a service fee from every valid transaction.

A 2% fee means:

\[
\text{processed amount} = \text{amount} - (\text{amount} \times 0.02)
\]

The function uses `map()` and a `lambda` expression as required.

```python
def apply_service_fee(amounts, fee_rate=0.02):
    processed = map(
        lambda amount: amount - (amount * fee_rate),
        amounts
    )
    return list(processed)
```

```python
processed_transactions = apply_service_fee(cleaned_transactions)

print("Cleaned transactions:")
print(cleaned_transactions)

print("\nTransactions after 2% service fee:")
print(processed_transactions)
```

## Task 2 - Recursive Nested Total

The function `calculate_nested_total(data)` follows the required recursive structure.

### Base cases
1. If `data` is `[]`, return `0.0`.
2. If `data` is an `int` or `float`, return the numeric value.

### Recursive step
For a non-empty list:

- `head = data[0]`
- `tail = data[1:]`

The function recursively calculates the total of both pieces and adds them together.

No `for` or `while` loop is used inside the recursive function.

```python
def calculate_nested_total(data):
    # Base case 1: empty list
    if data == []:
        return 0.0

    # Base case 2: single numeric value
    if isinstance(data, (int, float)):
        return float(data)

    # Recursive case: non-empty list
    head = data[0]
    tail = data[1:]

    return calculate_nested_total(head) + calculate_nested_total(tail)
```

## Test the Recursive Function

The following batch contains several levels of nested sub-transactions, including an empty list.

```python
nested_transactions = [
    100,
    [50, 25],
    [[10, 5], 40],
    [],
    [200, [125, [75, 50]]]
]

print("Nested transaction batch:")
print(nested_transactions)

nested_total = calculate_nested_total(nested_transactions)

print("\nTotal nested transaction value:")
print(nested_total)
```

## Additional Recursive Tests

These checks demonstrate that the same recursive function works with:

- a single number,
- an empty list,
- a flat list,
- and a deeply nested list.

```python
print("Single number:")
print(calculate_nested_total(150))

print("\nEmpty list:")
print(calculate_nested_total([]))

print("\nSimple list:")
print(calculate_nested_total([10, 20, 30]))

print("\nDeeply nested list:")
print(calculate_nested_total([10, [20, [30, [40]]]]))
```
